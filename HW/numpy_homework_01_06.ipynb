
### ✏️ Завдання 3.1

Є масив цін товарів у доларах. Виконай **без циклів**:

```python
prices_usd = np.array([15.99, 42.50, 8.00, 125.00, 33.75, 67.20, 9.99])
```

1. Переведи ціни в гривні (курс: **1 USD = 41.5 UAH**), результат округли до 2 знаків
2. Нарахуй ПДВ 20% — помнож гривневі ціни на 1.2, округли до 2 знаків
3. Знайди: мінімальну, максимальну, середню ціну з ПДВ
4. Знайди суму всіх товарів з ПДВ — це загальна вартість кошика
5. Виведи результати з поясненнями

````python
prices_usd = np.array([15.99, 42.50, 8.00, 125.00, 33.75, 67.20, 9.99])

# Твій код:
import numpy as np

# Курс долара
rate = 41.5

# 1. Переведення в гривні
prices_uah = np.round(prices_usd * rate, 2)

# 2. Нарахування ПДВ 20%
prices_vat = np.round(prices_uah * 1.2, 2)

# 3. Мінімальна, максимальна, середня ціна
min_price = prices_vat.min()
max_price = prices_vat.max()
avg_price = round(prices_vat.mean(), 2)

# 4. Загальна сума кошика
total = round(prices_vat.sum(), 2)

# Виведення результатів
print("Ціни у доларах:",prices_usd)
print("\n1.Ціни у гривнях:",prices_uah)
print("\n2.Ціни з ПДВ 20%:",prices_vat)


print("\n3.1. Мінімальна ціна:", min_price)
print("\n3.2. Максимальна ціна:", max_price)
print("\n3.3. Середня ціна:", avg_price)
#print("\n3.3. Середня ціна:", round(prices_vat.mean(), 2))
print("\n4. Загальна вартість кошика:", total)
````

Є масив часу виконання замовлень у хвилинах (норма — до 30 хв):

```python
delivery_time = np.array([22, 35, 28, 41, 19, 55, 27, 33, 24, 48,
                           31, 26, 52, 18, 29, 38, 23, 44, 20, 36])
```

1. Знайди всі замовлення, виконані **вчасно** (≤ 30 хв)
2. Знайди всі **запізнілі** замовлення (> 30 хв)
3. Порахуй кількість вчасних і запізнілих
4. Знайди відсоток запізнілих замовлень (округли до 1 знаку)
5. Яка середня затримка серед запізнілих замовлень?

````python
delivery_time = np.array([22, 35, 28, 41, 19, 55, 27, 33, 24, 48,
                           31, 26, 52, 18, 29, 38, 23, 44, 20, 36])

# Твій код:
#  всі замовлення, виконані вчасно (≤ 30 хв)
on_time = delivery_time[delivery_time <= 30]
print("1.Всі замовлення, виконані вчасно (≤ 30 хв):\n",on_time)

#  всі запізнілі замовлення (> 30 хв)
out_time = delivery_time[delivery_time > 30]
print("\n2.Всі запізнілі замовлення (> 30 хв):\n",out_time)

# Kількість вчасних
on_time_count = on_time.size
print("\n3.Kількість вчасних :",on_time_count)

# Kількість запізнілих
out_time_count = out_time.size
print("\n3.1. Kількість запізнілих:",out_time_count)

# відсоток запізнілих замовлень (округли до 1 знаку)
out_time_percent = round((out_time_count / delivery_time.size) * 100, 1)
print("\n4.Bідсоток запізнілих замовлень: ",out_time_percent,"%" )

# На скільки хвилин запізнювалися замовлення
delay = out_time - 30
print(" На скільки хвилин запізнювалися замовлення? ",delay,"хвилин")

# середня затримка серед запізнілих замовлень
avg_out_time = round(delay.mean(), 1)
print("\n5.Cередня затримка серед запізнілих замовлень:",avg_out_time,'хвилин')


````

3. За допомогою `np.where()` створи масив `status`:
   - `'Гарний день'` — якщо продажі вище середнього
   - `'Звичайний день'` — якщо нижче або рівно

````python


# Твій код:
import numpy as np

# Генерація випадкових продажів
np.random.seed(42)

# Продажі за місяць
daily = np.random.randint(100, 900, size=30)
print("\n1.Продажі за місяць:",daily)

# Середній продаж за місяць
average = daily.mean()
print("\n2.Середній продаж за місяць:",round(average, 2))

# 1. Дні з продажами вище середнього
above_avg = daily[daily > average]
print("\n3.Продажі вище середнього:",above_avg)

# 2. Кількість таких днів
count_above = above_avg.size
print("\n4.Кількість днів вище середнього:",count_above)

# 3. Створення масиву статусів
status = np.where(
    daily > average,
    'Гарний день',
    'Звичайний день'
)
print("\nПерші 10 днів:")
for i in range(10):
    print(f"День {i+1}: {daily[i]} -> {status[i]}")

````

### Запитання 3 — Фільтрація та маскування (Розділ 4)
1. Знайди кількість **«слабких» днів** — продажі нижче 400 грн
2. Знайди кількість **«сильних» днів** — продажі вище 700 грн
3. Яка середня виручка у слабкі дні і яка у сильні?
4. Створи масив `labels` за допомогою `np.where()`:
   - `'Слабкий'` якщо < 400
   - `'Сильний'` якщо > 700
   - `'Нормальний'` в решті випадків (підказка: вкладений `np.where`)
5. Виведи перші 14 днів: продажі та їх мітки

````python
# Твій код:
import numpy as np

# # Дані — не змінюй
# np.random.seed(2024)

# base = np.random.randint(200, 800, 56)

# weekend_boost = np.array([
#     1 if i % 7 in [5, 6] else 0
#     for i in range(56)
# ])

# sales_56 = (
#     base + weekend_boost * np.random.randint(100, 300, 56)
# ).astype(int)

# 1. Слабкі та сильні дні
weak_days = sales_56 < 400
strong_days = sales_56 > 700

# Кількість
weak_count = np.sum(weak_days)
strong_count = np.sum(strong_days)

print(f"Кількість слабких днів: {weak_count}")

print(f"Кількість сильних днів: {strong_count}")

# 2. Середня виручка
weak_average = sales_56[weak_days].mean()
strong_average = sales_56[strong_days].mean()

print(f"\nСередня виручка у слабкі дні: {weak_average:.2f}")

print(f"Середня виручка у сильні дні: {strong_average:.2f}")

# 3. Створення labels через np.where()
labels = np.where(sales_56 < 400,'Слабкий',np.where( sales_56 > 700,
 'Сильний','Нормальний'))

# 4. Перші 14 днів
print("\nПерші 14 днів:")

for i in range(14):
    print(
        f"День {i}: продажі = {sales_56[i]}, мітка = {labels[i]}"
    )
````
