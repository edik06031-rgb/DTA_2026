````python
my_dict = {
    1: "one",
    2: "two",
    3:"three",
    4: "four",
    5:"five"
}
````

NoneType - нам показали результат (або ні) у будь якому випадку, працювати з результатом повернення функціі ми не можемо

````python
type(None)
````

````python
# Повертає NoneType: без зміни даних
def info(d):
    print(d)
    # info(my_dict)
type(info(my_dict))
````

````python
# Повертає NoneType: змінює початкові дані
def info(d):
    d[6]= "six"
    print(d)
    # info(my_dict)
type(info(my_dict))
````

````python
# Повертає NoneType: змінює початкові дані
def info(d):
    d[6]= "six"
print(f"my_dict : {my_dict}")
    # info(my_dict)
type(info(my_dict))
````
