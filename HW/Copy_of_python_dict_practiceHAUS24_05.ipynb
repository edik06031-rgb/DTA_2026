# Розділ 1. Словники (`dict`)

**Швидке нагадування:**

Словник — це колекція пар `ключ : значення`, створюється у фігурних дужках:
```python
student = {"name": "Олена", "age": 20, "city": "Київ"}
```

**Основні операції:**
- Доступ: `student["name"]`
- Додавання/зміна: `student["email"] = "olena@example.com"`
- Видалення: `del student["age"]`
- Перевірка: `if "name" in student:`
- Безпечне отримання: `student.get("phone")` — повертає `None`, якщо ключа немає
- Довжина: `len(student)` — кількість пар

**Перебір у циклі:**
- `for key in d.keys():` — перебір ключів
- `for value in d.values():` — перебір значень
- `for key, value in d.items():` — перебір пар

### Завдання 20. Інвентар магазину ⭐⭐⭐

Дано словник з товарами магазину, де значення — це інший словник з ціною і кількістю на складі. Виконайте:
1. Виведіть загальну вартість всього товару на складі (ціна × кількість для кожного).
2. Виведіть назви товарів, яких залишилось менше 20 штук.
3. Виведіть найдорожчий товар.

````python
inventory = {
    "Ноутбук":   {"price": 18500, "stock": 12},
    "Телефон":   {"price": 9200,  "stock": 35},
    "Планшет":   {"price": 7800,  "stock": 8},
    "Навушники": {"price": 850,   "stock": 120},
    "Клавіатура":{"price": 1200,  "stock": 45},
    "Мишка":     {"price": 450,   "stock": 18}
}
# Ваш код:
total_value = 0
low_stock = []

max_name = ""
max_price = -1

for name, data in inventory.items():
    price = data["price"]
    stock = data["stock"]

    total_value += price * stock

    if stock < 20:
        low_stock.append(name)

    if price > max_price:
        max_price = price
        max_name = name

print("Загальна вартість товару на складі:", total_value, "грн")
print("Мало на складі (<20):", low_stock)
print("Найдорожчий товар:", max_name, "—", max_price, "грн")
````

````python
grades = {
    "Олена":   [85, 90, 78, 92, 88],
    "Андрій":  [72, 65, 80, 75, 70],
    "Марія":   [95, 92, 88, 90, 94],
    "Богдан":  [60, 65, 70, 55, 68],
    "Софія":   [82, 85, 80, 88, 84],
    "Петро":   [78, 75, 82, 80, 76],
    "Дарʼя":   [90, 88, 92, 85, 91]
}
# Ваш код:
averages = {}

# Середній бал кожного студента
for student, marks in grades.items():
    average = sum(marks) / len(marks)
    averages[student] = average

    print(student, "- середній бал:", average)

# Загальний середній бал групи
group_average = sum(averages.values()) / len(averages)

print("\nЗагальний середній бал групи:", group_average)

# Кількість студентів вище середнього
count = 0

for average in averages.values():
    if average > group_average:
        count += 1

print("Кількість студентів вище середнього:", count)

# Студент з найвищим середнім балом
best_student = max(averages, key=averages.get)

print("Студент з найвищим середнім балом:", best_student)
print("Його середній бал:", averages[best_student])
````

### Завдання 35. Рейтинг команд ⭐⭐

Дано список матчів. Кожен матч — це словник `{"team_a": ..., "team_b": ..., "score_a": ..., "score_b": ...}`. Створіть таблицю очок: за перемогу — 3 очка, за нічию — 1, за поразку — 0. Виведіть команди у порядку від найбільшої до найменшої кількості очок.

> 💡 **Підказка:** Створіть словник `points = {}`. У циклі по матчах: визначайте переможця і додавайте очки. Не забудьте обробити випадок, коли команди ще немає у словнику.

````python
matches = [
    {"team_a": "Динамо",     "team_b": "Шахтар",     "score_a": 1, "score_b": 2},
    {"team_a": "Зоря",       "team_b": "Динамо",     "score_a": 0, "score_b": 1},
    {"team_a": "Шахтар",     "team_b": "Дніпро",     "score_a": 3, "score_b": 1},
    {"team_a": "Динамо",     "team_b": "Ворскла",    "score_a": 2, "score_b": 2},
    {"team_a": "Шахтар",     "team_b": "Зоря",       "score_a": 4, "score_b": 0},
    {"team_a": "Дніпро",     "team_b": "Ворскла",    "score_a": 1, "score_b": 1},
    {"team_a": "Зоря",       "team_b": "Дніпро",     "score_a": 2, "score_b": 1},
    {"team_a": "Ворскла",    "team_b": "Шахтар",     "score_a": 0, "score_b": 3}
]
# Ваш код

points = {}

for match in matches:

    team_a = match["team_a"]
    team_b = match["team_b"]

    score_a = match["score_a"]
    score_b = match["score_b"]

    # Якщо команди ще немає у словнику
    if team_a not in points:
        points[team_a] = 0

    if team_b not in points:
        points[team_b] = 0

    # Перемога команди A
    if score_a > score_b:
        points[team_a] += 3

    # Перемога команди B
    elif score_b > score_a:
        points[team_b] += 3

    # Нічия
    else:
        points[team_a] += 1
        points[team_b] += 1

# Сортування за очками
sorted_teams = sorted(points.items(),
                      key=lambda x: x[1],
                      reverse=True)

print("Рейтинг команд:\n")

for team, pts in sorted_teams:
    print(team, "-", pts, "очок")
````

```python

```

> 💡 **Підказка:** Підрахуйте перемоги в окремих змінних. Цикл працює, поки обидва лічильники < 3. Правила перемоги: камінь б'є ножиці, ножиці — папір, папір — камінь.

````python
# Ваш код для завдання 41:
import random

user_score = 0
computer_score = 0

choices = ["камінь", "ножиці", "папір"]

while user_score < 3 and computer_score < 3:

    user_choice = input("Введіть камінь, ножиці або папір: ")

    if user_choice not in choices:
        print("Неправильний вибір!")
        continue

    computer_choice = random.choice(choices)

    print("Комп'ютер обрав:", computer_choice)

    # Нічия
    if user_choice == computer_choice:
        print("Нічия!")

    # Перемога користувача
    elif (
        (user_choice == "камінь" and computer_choice == "ножиці") or
        (user_choice == "ножиці" and computer_choice == "папір") or
        (user_choice == "папір" and computer_choice == "камінь")
    ):
        print("Ви виграли раунд!")
        user_score += 1

    # Перемога комп'ютера
    else:
        print("Комп'ютер виграв раунд!")
        computer_score += 1

    print("Рахунок:",
          user_score, "-", computer_score)

# Фінальний результат
print("\nГру завершено!")

if user_score == 3:
    print("Ви перемогли!")
else:
    print("Комп'ютер переміг!")
````

Створіть змінну `balance = 5000`. У циклі `while` запитуйте у користувача дію:
- `баланс` — показати баланс
- `зняти` — запитати суму і зняти (якщо вистачає коштів)
- `поповнити` — запитати суму і додати до балансу
- `історія` — показати список усіх операцій
- `вихід` — завершити

Зберігайте кожну операцію у списку як словник `{"type": "зняти", "amount": 500}`.

> 💡 **Підказка:** Створіть `history = []` перед циклом. Кожну операцію додавайте через `.append()`.

````python
# Ваш код для завдання 45:
balance = 5000
history = []

while True:
    action = input("\nВведіть дію (баланс / зняти / поповнити / історія / вихід): ").lower()

    # 1. Показати баланс
    if action == "баланс":
        print("Ваш баланс:", balance)
        history.append({"type": "баланс", "amount": 0})

    # 2. Зняти гроші
    elif action == "зняти":
        amount = int(input("Введіть суму для зняття: "))

        if amount <= balance:
            balance -= amount
            print(f"Знято {amount}. Новий баланс: {balance}")
            history.append({"type": "зняти", "amount": amount})
        else:
            print("Недостатньо коштів!")

    # 3. Поповнити рахунок
    elif action == "поповнити":
        amount = int(input("Введіть суму для поповнення: "))
        balance += amount
        print(f"Рахунок поповнено на {amount}. Баланс: {balance}")
        history.append({"type": "поповнити", "amount": amount})

    # 4. Історія операцій
    elif action == "історія":
        print("\nІсторія операцій:")
        for op in history:
            print(op)

    # 5. Вихід
    elif action == "вихід":
        print("Завершення роботи. Дякуємо!")
        break

    else:
        print("Невідома команда. Спробуйте ще раз.")
````

### Завдання 46. Менеджер задач (todo) ⭐⭐⭐

Створіть програму-менеджер задач. Задачі зберігаються у списку словників: `{"task": "...", "done": False}`. У циклі `while` запитуйте дію:
- `додати` — додати нову задачу
- `показати` — показати всі задачі з номерами і статусом
- `виконати` — за номером позначити задачу як виконану
- `видалити` — за номером видалити задачу
- `вихід` — завершити

````python
# Ваш код для завдання 46:
tasks = []

while True:
    action = input("\nВведіть дію (додати / показати / виконати / видалити / вихід): ").lower()

    # 1. Додати задачу
    if action == "додати":
        task_text = input("Введіть задачу: ")
        tasks.append({"task": task_text, "done": False})
        print("Задачу додано!")

    # 2. Показати задачі
    elif action == "показати":
        if not tasks:
            print("Список задач порожній.")
        else:
            for i, t in enumerate(tasks, start=1):
                status = "✔" if t["done"] else "✘"
                print(f"{i}. [{status}] {t['task']}")

    # 3. Виконати задачі
    elif action == "виконати":
        if not tasks:
            print("Немає задач.")
        else:
            num = int(input("Введіть номер задачі: "))
            if 1 <= num <= len(tasks):
                tasks[num - 1]["done"] = True
                print("Задачу позначено як виконану ✔")
            else:
                print("Невірний номер!")

    # 4. Видалити задачу
    elif action == "видалити":
        if not tasks:
            print("Немає задач.")
        else:
            num = int(input("Введіть номер задачі: "))
            if 1 <= num <= len(tasks):
                removed = tasks.pop(num - 1)
                print(f"Видалено: {removed['task']}")
            else:
                print("Невірний номер!")

    # 5. Вихід
    elif action == "вихід":
        print("Завершення програми.")
        break

    else:
        print("Невідома команда.")
````

> 💡 **Підказка:** Зберігайте всі використані слова у списку. Першу і останню літеру отримуйте через індекси: `word[0]` і `word[-1]`.

````python
# Ваш код для завдання 52:
used_words = []

current_word = input("Введіть стартове слово: ").lower()
used_words.append(current_word)

while True:
    next_word = input("Наступне слово (або 'здаюсь'): ").lower()

    # 1. вихід з гри
    if next_word == "здаюсь":
        print("Гру завершено. Ви здались 😄")
        break

    # 2. перевірка повтору
    if next_word in used_words:
        print(" Це слово вже використовувалось! Ви програли.")
        break

    # 3. перевірка першої літери
    if next_word[0] != current_word[-1]:
        print(f" Потрібно слово на '{current_word[-1]}'! Ви програли.")
        break

    # 4. якщо все ок
    used_words.append(next_word)
    current_word = next_word
    print("✔ Добре! Продовжуємо гру.")
````

2. У яку годину було найбільше ERROR-ів
3. Усі повідомлення з рівнем ERROR
4. Чи були години без жодного ERROR (з тих, у яких є записи)

````python
logs = [
    {"level": "INFO",    "message": "Сервер запущено",     "hour": 8},
    {"level": "INFO",    "message": "Користувач увійшов", "hour": 9},
    {"level": "WARNING", "message": "Повільний запит",     "hour": 9},
    {"level": "ERROR",   "message": "Помилка БД",          "hour": 10},
    {"level": "ERROR",   "message": "Тайм-аут запиту",     "hour": 10},
    {"level": "INFO",    "message": "Запит виконано",      "hour": 11},
    {"level": "WARNING", "message": "Низько місця на диску","hour": 12},
    {"level": "ERROR",   "message": "Помилка БД",          "hour": 14},
    {"level": "INFO",    "message": "Резервна копія",      "hour": 15},
    {"level": "ERROR",   "message": "Збій авторизації",    "hour": 14},
    {"level": "ERROR",   "message": "Помилка БД",          "hour": 14},
    {"level": "INFO",    "message": "Користувач вийшов",   "hour": 17}
]
# Ваш код:




level_count = {}        # 1. Кількість записів кожного рівня

error_by_hour = {}      # 2. ERROR по годинах

error_messages = []     # 3. Усі ERROR повідомлення


for log in logs:             # обхід логів
    level = log["level"]
    hour = log["hour"]
    message = log["message"]

    level_count[level] = level_count.get(level, 0) + 1    # підрахунок рівнів


    if level == "ERROR":                                  # ERROR логіка
        error_messages.append(message)
        error_by_hour[hour] = error_by_hour.get(hour, 0) + 1



most_error_hour = max(error_by_hour, key=error_by_hour.get)   # година з найбільшою кількістю ERROR




all_hours = set(log["hour"] for log in logs)      #  чи були години без ERROR
error_hours = set(error_by_hour.keys())

hours_without_error = all_hours - error_hours


print(" Рівні логів:", level_count)
print(" Година з найбільшою кількістю ERROR:", most_error_hour)
print(" ERROR повідомлення:", error_messages)
print(" Години без ERROR:", sorted(hours_without_error))
````
